## **List Comprehensions**

```python
"""
01 - LIST COMPREHENSION
=======================
WHY: A for-loop that builds a list takes 3-4 lines. A list comprehension does it in 1,
and it is faster (no repeated .append() lookups) and easier to read once you know the pattern.

SYNTAX:
    [expression  for item in iterable  if condition]
     ^ what to    ^ loop               ^ optional filter
       keep

READ IT AS: "give me <expression> for every <item> in <iterable> (only if <condition>)"
Run:  python 01_list_comprehension.py
"""

# ---------- 1. LOOP -> COMPREHENSION ----------
squares_loop = []
for n in range(1, 6):
    squares_loop.append(n ** 2)

squares = [n ** 2 for n in range(1, 6)]
print("1.", squares)                                  # [1, 4, 9, 16, 25]
assert squares == squares_loop

# ---------- 2. FILTER (if at the END) ----------
evens = [n for n in range(10) if n % 2 == 0]
print("2.", evens)                                    # [0, 2, 4, 6, 8]

# ---------- 3. IF-ELSE (goes at the FRONT, before 'for') ----------
# RULE: filter only  -> 'if' at the end
#       choose value -> 'if/else' at the front
labels = ["even" if n % 2 == 0 else "odd" for n in range(5)]
print("3.", labels)                                   # ['even','odd','even','odd','even']

# ---------- 4. NESTED LOOPS ----------
# Order is the SAME as nested for-loops written top to bottom.
pairs = [(x, y) for x in [1, 2] for y in ["a", "b"]]
print("4.", pairs)                                    # [(1,'a'),(1,'b'),(2,'a'),(2,'b')]

# ---------- 5. FLATTEN a 2D list ----------
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flat = [num for row in matrix for num in row]
print("5.", flat)                                     # [1..9]

# ---------- 6. TRANSPOSE a matrix (nested comprehension) ----------
transposed = [[row[i] for row in matrix] for i in range(3)]
print("6.", transposed)                               # [[1,4,7],[2,5,8],[3,6,9]]

# =====================================================
# REAL-WORLD SCENARIOS
# =====================================================

# --- A. Clean user input (forms, CSV, scraped data) ---
raw_emails = ["  Ali@Gmail.com ", "SARA@yahoo.com", "", "  ", "bob@outlook.com "]
clean_emails = [e.strip().lower() for e in raw_emails if e.strip()]
print("A.", clean_emails)

# --- B. Extract one field from API / JSON records ---
users = [
    {"id": 1, "name": "Ali",  "active": True,  "age": 22},
    {"id": 2, "name": "Sara", "active": False, "age": 30},
    {"id": 3, "name": "Bilal","active": True,  "age": 17},
]
active_names = [u["name"] for u in users if u["active"]]
print("B.", active_names)                             # ['Ali', 'Bilal']

# --- C. Filter log lines ---
logs = [
    "INFO  server started",
    "ERROR db connection failed",
    "INFO  request /home",
    "ERROR timeout on /api",
]
errors = [line for line in logs if line.startswith("ERROR")]
print("C.", errors)

# --- D. Apply a function to every item (price + tax) ---
prices = [100, 250, 399.5]
with_tax = [round(p * 1.18, 2) for p in prices]
print("D.", with_tax)

# --- E. Convert types (very common: strings from CSV -> numbers) ---
row = ["10", "20", "x", "40"]
nums = [int(v) for v in row if v.isdigit()]
print("E.", nums)                                     # [10, 20, 40]

# --- F. Call a function inside (e.g. build file paths / URLs) ---
ids = [101, 102, 103]
urls = [f"https://api.example.com/users/{i}" for i in ids]
print("F.", urls)

# --- G. Keep the whole object, filter by condition ---
adults = [u for u in users if u["age"] >= 18]
print("G.", [u["name"] for u in adults])

# --- H. Normalise words (NLP preprocessing) ---
sentence = "The Quick brown FOX, jumps!"
tokens = [w.strip(",.!").lower() for w in sentence.split()]
print("H.", tokens)

# =====================================================
# WHEN NOT TO USE
# =====================================================
# - More than ~2 'for' clauses or complex logic -> use a normal loop (readability wins).
# - You only need side effects (print, save to DB) -> use a for-loop, not a comprehension.
# - Huge data you only loop over once -> use a generator expression (see file 04).

# =====================================================
# PRACTICE (solve first, then check the asserts)
# =====================================================
# Q1: cubes of numbers 1..5
# Q2: only words longer than 3 letters, uppercased, from ["a","tree","is","green"]
# Q3: replace negatives with 0 in [3,-1,4,-5]
# Q4: all (x, y) with x in 1..3, y in 1..3, where x != y

q1 = [n ** 3 for n in range(1, 6)]
q2 = [w.upper() for w in ["a", "tree", "is", "green"] if len(w) > 3]
q3 = [n if n > 0 else 0 for n in [3, -1, 4, -5]]
q4 = [(x, y) for x in range(1, 4) for y in range(1, 4) if x != y]

assert q1 == [1, 8, 27, 64, 125]
assert q2 == ["TREE", "GREEN"]
assert q3 == [3, 0, 4, 0]
assert len(q4) == 6
print("All practice answers correct")
```

## **Dictionary Comprehension**

```python
"""
02 - DICTIONARY COMPREHENSION
=============================
WHY: Dicts are lookup tables. Dict comprehensions build, reshape, invert and filter
them in one line. You use this constantly with JSON, configs, and databases.

SYNTAX:
    {key_expr: value_expr  for item in iterable  if condition}

Note the curly braces AND the colon  key: value.
Run:  python 02_dict_comprehension.py
"""

# ---------- 1. BASIC ----------
squares = {n: n ** 2 for n in range(1, 6)}
print("1.", squares)                                  # {1:1, 2:4, 3:9, 4:16, 5:25}

# ---------- 2. FROM TWO LISTS (zip) ----------
names = ["Ali", "Sara", "Bilal"]
marks = [88, 92, 75]
result = {n: m for n, m in zip(names, marks)}         # same as dict(zip(names, marks))
print("2.", result)

# ---------- 3. FILTER ----------
passed = {n: m for n, m in result.items() if m >= 80}
print("3.", passed)                                   # {'Ali': 88, 'Sara': 92}

# ---------- 4. TRANSFORM VALUES ----------
curved = {n: min(m + 5, 100) for n, m in result.items()}
print("4.", curved)

# ---------- 5. INVERT a dict (swap key <-> value) ----------
# Only safe when values are unique!
inverted = {v: k for k, v in result.items()}
print("5.", inverted)

# ---------- 6. IF-ELSE on value ----------
grades = {n: ("pass" if m >= 80 else "fail") for n, m in result.items()}
print("6.", grades)


# =====================================================
# REAL-WORLD SCENARIOS
# =====================================================

# --- A. Index a list of records by ID (fastest way to look things up) ---
users = [
    {"id": 1, "name": "Ali",  "city": "Karachi"},
    {"id": 2, "name": "Sara", "city": "Lahore"},
    {"id": 3, "name": "Bilal","city": "Karachi"},
]
by_id = {u["id"]: u for u in users}
print("A.", by_id[2]["name"])                         # O(1) lookup instead of looping

# --- B. Count frequencies (word count, category count) ---
words = ["apple", "bat", "apple", "cat", "bat", "apple"]
counts = {w: words.count(w) for w in set(words)}
print("B.", counts)
# For big data use collections.Counter - but this shows the pattern.

# --- C. Clean/normalise keys (messy CSV headers, API fields) ---
raw_row = {" Name ": "Ali", "AGE": "22", " City": "Karachi"}
clean_row = {k.strip().lower(): v.strip() for k, v in raw_row.items()}
print("C.", clean_row)

# --- D. Load config / env vars with a prefix ---
env = {"APP_HOST": "localhost", "APP_PORT": "8000", "PATH": "/usr/bin", "APP_DEBUG": "1"}
config = {k[4:].lower(): v for k, v in env.items() if k.startswith("APP_")}
print("D.", config)                                   # {'host':..., 'port':..., 'debug':...}

# --- E. Pick only some fields from a record (hide password, etc.) ---
user = {"id": 1, "name": "Ali", "email": "a@x.com", "password": "secret"}
safe_fields = {"id", "name", "email"}
public = {k: v for k, v in user.items() if k in safe_fields}
print("E.", public)

# --- F. Group data (dict of lists) ---
by_city = {
    city: [u["name"] for u in users if u["city"] == city]
    for city in {u["city"] for u in users}
}
print("F.", by_city)                                  # {'Karachi':['Ali','Bilal'], 'Lahore':['Sara']}

# --- G. Convert types across a whole dict ---
form = {"age": "22", "marks": "88", "name": "Ali"}
parsed = {k: int(v) if v.isdigit() else v for k, v in form.items()}
print("G.", parsed)

# --- H. Mapping for ML: label <-> id (every NLP/classification project) ---
labels = ["angry", "happy", "sad"]
label2id = {lab: i for i, lab in enumerate(labels)}
id2label = {i: lab for lab, i in label2id.items()}
print("H.", label2id, id2label)


# =====================================================
# GOTCHAS
# =====================================================
# - Duplicate keys: the LAST one wins silently.
# - Don't invert dicts with duplicate values (data is lost).
# - Keys must be hashable (str, int, tuple OK; list NOT OK).


# =====================================================
# PRACTICE
# =====================================================
# Q1: {1:'1', 2:'2', 3:'3'}  (number -> its string)
# Q2: from {"a":1,"b":2,"c":3} keep only values > 1
# Q3: {word: length} for ["python","is","fun"]
# Q4: double every price in {"tea":50,"cake":120}

q1 = {n: str(n) for n in range(1, 4)}
q2 = {k: v for k, v in {"a": 1, "b": 2, "c": 3}.items() if v > 1}
q3 = {w: len(w) for w in ["python", "is", "fun"]}
q4 = {k: v * 2 for k, v in {"tea": 50, "cake": 120}.items()}

assert q1 == {1: "1", 2: "2", 3: "3"}
assert q2 == {"b": 2, "c": 3}
assert q3 == {"python": 6, "is": 2, "fun": 3}
assert q4 == {"tea": 100, "cake": 240}
print("All practice answers correct")
```

## **Set comprehensions**

```python
"""
03 - SET COMPREHENSION
======================
WHY: A set stores only UNIQUE items and checks membership instantly.
Use a set comprehension whenever you need "unique values of something".

SYNTAX:
    {expression  for item in iterable  if condition}

Curly braces but NO colon (colon = dict, no colon = set).
WARNING: {} alone is an empty DICT. Empty set is set().
Sets are UNORDERED - don't rely on print order.
Run:  python 03_set_comprehension.py
"""

# ---------- 1. BASIC (duplicates removed automatically) ----------
nums = [1, 2, 2, 3, 3, 3, 4]
unique_squares = {n ** 2 for n in nums}
print("1.", sorted(unique_squares))                   # [1, 4, 9, 16]

# ---------- 2. FILTER ----------
unique_evens = {n for n in nums if n % 2 == 0}
print("2.", unique_evens)                             # {2, 4}

# ---------- 3. NORMALISE then dedupe ----------
names = ["Ali", "ali", "ALI", "Sara", "sara"]
unique_names = {n.lower() for n in names}
print("3.", sorted(unique_names))                     # ['ali', 'sara']

# ---------- 4. SET OPERATIONS (the real power) ----------
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print("4.", a | b, a & b, a - b, a ^ b)
#   union | intersection & | difference - | symmetric diff ^


# =====================================================
# REAL-WORLD SCENARIOS
# =====================================================

# --- A. Unique domains from a list of emails ---
emails = ["ali@gmail.com", "sara@yahoo.com", "bob@gmail.com", "zoya@gmail.com"]
domains = {e.split("@")[1] for e in emails}
print("A.", sorted(domains))                          # ['gmail.com', 'yahoo.com']

# --- B. Unique categories / tags from records ---
products = [
    {"name": "Phone",  "category": "electronics"},
    {"name": "Laptop", "category": "electronics"},
    {"name": "Shirt",  "category": "clothing"},
]
categories = {p["category"] for p in products}
print("B.", sorted(categories))

# --- C. Find NEW leads (not already in your database) - lead-gen style ---
existing_leads = {"a@x.com", "b@x.com", "c@x.com"}
scraped = ["B@x.com ", "d@x.com", "E@x.com", "a@x.com"]
scraped_clean = {e.strip().lower() for e in scraped}
new_leads = scraped_clean - existing_leads            # set difference
print("C.", sorted(new_leads))                        # ['d@x.com', 'e@x.com']

# --- D. Fast membership check (blacklist / allowlist) ---
blocked_ids = {u for u in range(1000, 1010)}
print("D.", 1005 in blocked_ids)                      # True - O(1), list would be O(n)

# --- E. Unique words in a text (vocabulary building for NLP) ---
text = "to be or not to be that is the question"
vocab = {w for w in text.split()}
print("E.", len(vocab), "unique words")

# --- F. Common interests between two users ---
ali = ["cricket", "coding", "gaming"]
sara = ["coding", "art", "gaming"]
common = {x for x in ali} & {x for x in sara}
print("F.", sorted(common))                           # ['coding', 'gaming']

# --- G. Unique first letters ---
print("G.", sorted({w[0] for w in ["apple", "avocado", "banana", "blueberry", "cherry"]}))

# --- H. Detect duplicates in a list ---
ids = [1, 2, 3, 2, 4, 1]
seen = set()
dups = {i for i in ids if i in seen or seen.add(i)}   # trick; for beginners prefer a loop
print("H.", dups)                                     # {1, 2}


# =====================================================
# GOTCHAS
# =====================================================
# - No order, no indexing: s[0] fails.
# - Items must be hashable (no lists/dicts inside a set). Use tuples instead.
# - To keep order while deduping use:  list(dict.fromkeys(items))


# =====================================================
# PRACTICE
# =====================================================
# Q1: unique lengths of words in ["hi","yo","hello","hey"]
# Q2: unique uppercase letters present in "Hello World"
# Q3: numbers in range(20) divisible by 3 OR 5
# Q4: emails in list1 that are NOT in list2

q1 = {len(w) for w in ["hi", "yo", "hello", "hey"]}
q2 = {c for c in "Hello World" if c.isupper()}
q3 = {n for n in range(20) if n % 3 == 0 or n % 5 == 0}
list1, list2 = ["a@x.com", "b@x.com", "c@x.com"], ["b@x.com"]
q4 = {e for e in list1} - {e for e in list2}

assert q1 == {2, 3, 5}
assert q2 == {"H", "W"}
assert q3 == {0, 3, 5, 6, 9, 10, 12, 15, 18}
assert q4 == {"a@x.com", "c@x.com"}
print("All practice answers correct")
```

## **Tuple Comprehensions**

```python
"""
04 - "TUPLE COMPREHENSION" (the truth)
=====================================
There is NO tuple comprehension in Python.

    (x for x in range(5))    <- this is a GENERATOR EXPRESSION, not a tuple!

Tuples are immutable, so Python gives you 2 real ways to get tuple results:

    1) tuple(x for x in ...)      -> builds an actual tuple
    2) (x for x in ...)           -> lazy generator (computes one item at a time)

WHY GENERATORS MATTER: they use almost zero memory. A list of 10 million items
takes hundreds of MB; a generator over the same data takes a few hundred BYTES.

SYNTAX:
    tuple(expression for item in iterable if condition)
    (expression for item in iterable if condition)      # generator

Run:  python 04_tuple_comprehension.py
"""
import sys

# ---------- 1. THE COMMON MISTAKE ----------
not_a_tuple = (x for x in range(5))
print("1.", type(not_a_tuple))                        # <class 'generator'>

# ---------- 2. REAL TUPLE ----------
real_tuple = tuple(x ** 2 for x in range(5))
print("2.", real_tuple, type(real_tuple))             # (0, 1, 4, 9, 16) <class 'tuple'>

# ---------- 3. GENERATOR IS LAZY + ONE-TIME USE ----------
gen = (x * 2 for x in range(3))
print("3.", next(gen), next(gen), next(gen))          # 0 2 4
print("   exhausted:", list(gen))                     # [] - already used up!

# ---------- 4. MEMORY PROOF ----------
big_list = [x for x in range(1_000_000)]
big_gen = (x for x in range(1_000_000))
print("4. list:", sys.getsizeof(big_list), "bytes | generator:", sys.getsizeof(big_gen), "bytes")


# =====================================================
# REAL-WORLD SCENARIOS
# =====================================================

# --- A. sum / min / max / any / all directly on a generator (no list needed) ---
orders = [
    {"id": 1, "amount": 500,  "paid": True},
    {"id": 2, "amount": 1200, "paid": False},
    {"id": 3, "amount": 300,  "paid": True},
]
total_paid = sum(o["amount"] for o in orders if o["paid"])
print("A.", total_paid)                               # 800

# --- B. any() / all() validation - stops early, very fast ---
has_unpaid = any(not o["paid"] for o in orders)
all_positive = all(o["amount"] > 0 for o in orders)
print("B.", has_unpaid, all_positive)                 # True True

# --- C. Process a HUGE file line by line without loading it ---
# lines = (line.strip() for line in open("huge.csv"))
# errors = (l for l in lines if "ERROR" in l)
# for e in errors: handle(e)
# Each stage is lazy - only one line is in memory at a time.
lines = (l.strip() for l in ["ok", "ERROR a", "ok", "ERROR b"])
errors = (l for l in lines if l.startswith("ERROR"))
print("C.", list(errors))

# --- D. Prepare rows for a database bulk insert (list of tuples) ---
users = [{"name": "Ali", "age": 22}, {"name": "Sara", "age": 30}]
rows = [(u["name"], u["age"]) for u in users]
print("D.", rows)
# cursor.executemany("INSERT INTO users VALUES (?, ?)", rows)   # sqlite3 expects tuples

# --- E. Immutable data you don't want changed (constants, dict keys) ---
RGB_LEVELS = tuple(v for v in range(0, 256, 85))
print("E.", RGB_LEVELS)                               # (0, 85, 170, 255)

# --- F. Coordinates / pairs as tuples (can be used as dict keys or in sets) ---
grid = {(x, y) for x in range(2) for y in range(2)}
print("F.", sorted(grid))                             # [(0,0),(0,1),(1,0),(1,1)]

# --- G. Find first match (stop as soon as found) ---
first_big = next((o for o in orders if o["amount"] > 1000), None)
print("G.", first_big)

# --- H. Chain generators (data pipeline) ---
raw = ["  10 ", "20", "bad", " 30"]
stripped = (s.strip() for s in raw)
valid = (s for s in stripped if s.isdigit())
numbers = (int(s) for s in valid)
print("H.", sum(numbers))                             # 60


# =====================================================
# WHEN TO USE WHAT
# =====================================================
#   Need to index / reuse / print results     -> list [...]
#   Looping once over big/streaming data      -> generator (...)
#   Need immutable result                     -> tuple(...)
#   Feeding sum/min/max/any/all/join          -> generator (drop the brackets)


# =====================================================
# PRACTICE
# =====================================================
# Q1: tuple of cubes of 1..4
# Q2: sum of squares of odd numbers below 10 (use generator)
# Q3: True if every word in a list has length > 2 (use all)
# Q4: first number in [3,8,12,5] greater than 10, else None (use next)

q1 = tuple(n ** 3 for n in range(1, 5))
q2 = sum(n ** 2 for n in range(10) if n % 2 == 1)
q3 = all(len(w) > 2 for w in ["cat", "tree", "sun"])
q4 = next((n for n in [3, 8, 12, 5] if n > 10), None)

assert q1 == (1, 8, 27, 64)
assert q2 == 165
assert q3 is True
assert q4 == 12
print("All practice answers correct")
```
